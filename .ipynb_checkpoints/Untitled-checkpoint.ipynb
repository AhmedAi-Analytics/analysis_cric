```python
df['country']=df['Player'].str.split(pat="(").str[1]
```

```python
df['country']=df['country'].str.split(pat=")").str[0]
```

```python
df.info()
```

```python
df['Highest_Score']=df['Highest_Score'].str.split(pat="*").str[0]
```

```python
df['Highest_Score']=df['Highest_Score'].astype(int)
```

```python
df.info()
```

```python
df
```

```python
df[df["0"]=="-"]
```

```python
df["Strike_Rate"]=df["Strike_Rate"].replace("-",0)
```

```python
df["Ball_Faced"]=df["Ball_Faced"].replace("-",0)
```

```python
df["0"]=df["0"].replace("-",0)
```

```python
df["Ball_Faced"]=df['Ball_Faced'].str.split(pat="+").str[0]
```

```python
df["Ball_Faced"]=df['Ball_Faced'].astype(float)
```
